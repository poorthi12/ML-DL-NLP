```python
stemming = ['eating','eaten','going','gone','walking','went','jumping']
```

```python
##PORTER STEMMER
from nltk import PorterStemmer
stemming_word = PorterStemmer()
for word in stemming:
    print(word + " -----> " + stemming_word.stem(word))
```

```python
#this kind of error is the problem
stemming_word.stem("congradulations")
```

```python
##RegexpStemmer class
from nltk import RegexpStemmer
reg_stem = RegexpStemmer('ing$|s$|e$|able$', min=4)
reg_stem.stem("comfortable")
```

```python
## the $ is used to tell from where it need to be removed because it is in last like playing it removes play(ing)  
from nltk import RegexpStemmer
reg_stem = RegexpStemmer('ing$|s$|e$|able$', min=4)
reg_stem.stem("playing")
```

```python
## if there is mo $ it will remove every ing in a word 
from nltk import RegexpStemmer
reg_stem = RegexpStemmer('ing|s|e|able|manish', min=5)
print(reg_stem.stem("ingplaying"))
print(reg_stem.stem("ablecomfortable"))
print(reg_stem.stem("scopys"))
print(reg_stem.stem("emane"))
print(reg_stem.stem("manishpaani"))
```

```python
##Snowball Stemmer
from nltk.stem import SnowballStemmer
snow = SnowballStemmer('english')
for words in stemming:
    print(words + " -----> " + snow.stem(words))
```

**The difference between porterStemmer and SnowballStemmer**

```python
print(stemming_word.stem("fairly"),stemming_word.stem("sportingly"))
print(snow.stem("fairly"),snow.stem("sportingly"))
##both are not accurate but snow is littel good
print(snow.stem('goes') + " ----> This is the problem in SnowBallStemmer")
```
