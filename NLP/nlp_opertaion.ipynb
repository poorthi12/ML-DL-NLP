```python
corpus = " Hello Welcome, i'm poorna here Studying NLP. I'm very intrested to study new topics! especially in AI models."
```

```python
from nltk import sent_tokenize
sent_tokenize(corpus)
```

```python
type(sent_tokenize(corpus))
```

```python
#since the type is list we can iterate soo..
document = sent_tokenize(corpus)
```

```python
for sentence in document:
    print(sentence)
```

```python
##Tokenization
##paragraph -->
##word -->
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenized = TreebankWordTokenizer()
tokenized.tokenize(corpus)
```
