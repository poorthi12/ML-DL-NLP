```python
import nltk
sentence = "The Eiffel Tower was built from 1887 to 1889 by French engineer Gustave Eiffel, whose company specialized in building metal frameworks and structures."
word = nltk.word_tokenize(sentence)
```

```python
print(word)
```

```python
import nltk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')

tag_ele = nltk.pos_tag(word)
print(tag_ele)
```

```python
nltk.download("words")
nltk.download("maxent_ne_chunker")
```

```python
# download the required NER chunker model
nltk.download("maxent_ne_chunker_tab", quiet=True)

# ne_chunk expects a list of (word, tag) tuples
ner_tree = nltk.ne_chunk(tag_ele)

# print it in Jupyter; use .draw() only if GUI display works in your environment
print(ner_tree)
# ner_tree.draw()
```

```python
tree = nltk.ne_chunk(tag_ele)
print(tree)
```
